```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

import warnings
warnings.filterwarnings('ignore')

# Membaca dataset hasil preprocessing
df = pd.read_csv('data/processed/supermarket_sales_clean.csv')

# Mengembalikan kolom Date ke format datetime
df['Date'] = pd.to_datetime(df['Date'])

print("Dataset berhasil dimuat.")
print("Ukuran dataset:", df.shape)

display(df.head())
```

```python
total_revenue = df['Total'].sum()
total_transactions = df['Invoice ID'].nunique()
total_quantity = df['Quantity'].sum()
average_rating = df['Rating'].mean()
average_transaction = df['Total'].mean()

print("RINGKASAN PENJUALAN SUPERMARKET")
print("=" * 40)
print(f"Total Revenue       : {total_revenue:,.2f}")
print(f"Total Transactions  : {total_transactions:,}")
print(f"Total Quantity      : {total_quantity:,}")
print(f"Average Rating      : {average_rating:.2f}")
print(f"Average Transaction : {average_transaction:,.2f}")
```

```python
sales_product = (
    df.groupby('Product line')['Total']
    .sum()
    .sort_values(ascending=False)
)

display(sales_product.to_frame(name='Total Sales'))
```

```python
sales_branch = (
    df.groupby(['Branch', 'City'])['Total']
    .sum()
    .sort_values(ascending=False)
)

display(sales_branch.to_frame(name='Total Sales'))
```

```python
payment_analysis = (
    df.groupby('Payment')
    .agg(
        Total_Transactions=('Invoice ID', 'nunique'),
        Total_Sales=('Total', 'sum')
    )
    .sort_values('Total_Sales', ascending=False)
)

display(payment_analysis)
```

```python
monthly_sales = (
    df.groupby('Month Name')['Total']
    .sum()
)

month_order = [
    'January',
    'February',
    'March'
]

monthly_sales = monthly_sales.reindex(month_order)

display(monthly_sales.to_frame(name='Total Sales'))
```
