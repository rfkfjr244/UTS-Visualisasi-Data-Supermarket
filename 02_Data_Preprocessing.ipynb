```python
import pandas as pd
import numpy as np
import os

# Memeriksa lokasi kerja
print("Lokasi kerja:", os.getcwd())

# Membaca dataset mentah
df = pd.read_csv('data/raw/supermarket_sales.csv')

# Menghapus kolom yang seluruh isinya kosong
df = df.dropna(axis=1, how='all')

# Membuat salinan untuk preprocessing
df_clean = df.copy()

print("\nData berhasil dimuat.")
print("Ukuran dataset:", df_clean.shape)

display(df_clean.head())
```

```python
# Mengubah kolom Date menjadi datetime
df_clean['Date'] = pd.to_datetime(
    df_clean['Date'],
    format='%m/%d/%Y',
    errors='coerce'
)

# Mengubah kolom Time menjadi datetime
df_clean['Time'] = pd.to_datetime(
    df_clean['Time'],
    format='%H:%M:%S',
    errors='coerce'
).dt.time

print("Tipe data setelah konversi:")
display(df_clean.dtypes)
```

```python
print("Tanggal gagal dikonversi:",
      df_clean['Date'].isnull().sum())

print("Waktu gagal dikonversi:",
      df_clean['Time'].isnull().sum())

display(df_clean[['Date', 'Time']].head())
```

```python
df_clean['Year'] = df_clean['Date'].dt.year
df_clean['Month'] = df_clean['Date'].dt.month
df_clean['Month Name'] = df_clean['Date'].dt.month_name()
df_clean['Day'] = df_clean['Date'].dt.day
df_clean['Day Name'] = df_clean['Date'].dt.day_name()

display(df_clean.head())
```

```python
print("Jumlah baris:", df_clean.shape[0])
print("Jumlah kolom:", df_clean.shape[1])

print("\nJumlah nilai kosong:")
display(df_clean.isnull().sum())

print("\nJumlah duplikasi:")
print(df_clean.duplicated().sum())
```

```python
import os

os.makedirs('data/processed', exist_ok=True)

df_clean.to_csv(
    'data/processed/supermarket_sales_clean.csv',
    index=False
)

print("Dataset hasil preprocessing berhasil disimpan!")
```
